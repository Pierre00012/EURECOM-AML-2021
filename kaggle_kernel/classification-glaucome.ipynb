```python
import os
import sys
import subprocess

GIT_SHA = "__GIT_SHA__"

REPO = "https://github.com/Pierre00012/EURECOM-AML-2021"
REPO_DIR = "/kaggle/working/repo"
```

```python
subprocess.run(
    ["git", "clone", REPO, REPO_DIR],
    check=True
)

subprocess.run(
    ["git", "checkout", GIT_SHA],
    cwd=REPO_DIR,
    check=True
)

sys.path.insert(0, REPO_DIR)
os.environ["PYTHONPATH"] = REPO_DIR
os.environ["GIT_SHA"] = GIT_SHA
```

```python
DATA_DIR = "/kaggle/input"

for root, dirs, files in os.walk(DATA_DIR):
    print(root, files[:5])
```

```python
subprocess.run(
    [
        sys.executable,
        "-m",
        "pip",
        "install",
        "-r",
        f"{REPO_DIR}/requirements.txt"
    ],
    check=True
)
```

```python
import os
from kaggle_secrets import UserSecretsClient

# Récupération du secret Kaggle
user_secrets = UserSecretsClient()

wandb_key = user_secrets.get_secret("WANDB_API_KEY")

# Transmission aux scripts Python
os.environ["WANDB_API_KEY"] = wandb_key

print("W&B API key configured successfully.")
```

```python
env = os.environ.copy()

subprocess.run(
    [sys.executable, f"{REPO_DIR}/src/train.py"],
    check=True,
    env=env
)

subprocess.run(
    [sys.executable, f"{REPO_DIR}/src/evaluate.py"],
    check=True,
    env=env
)
```
